````python
import os
import re
import json
import glob
import pandas as pd

# ----------------------------
# 1) Load CSV
# ----------------------------
df = pd.read_csv("all_questions_answers_scores.csv")

# Add "reasoning" column right after MedGPT5
insert_at = df.columns.get_loc("MedGPT5") + 1
if "reasoning" not in df.columns:
    df.insert(insert_at, "reasoning", pd.NA)

GOLD_MODEL_NAME = "Rephrased Gold Answers"

GOLD_BASE_DIR = (
    "/cluster/home/gcardenal/HIV/medical_llm_evaluation/"
    "deploy_medical_llm_evaluation/gold_standard_rephrasing/"
    "results-GPT-score-rephrased"
)

# ----------------------------
# 2) Helpers to parse scoring files
# ----------------------------
FENCE_RE = re.compile(r"```(?:json)?\s*(\{.*?\})\s*```", re.DOTALL | re.IGNORECASE)

def extract_json_from_response(text: str) -> dict | None:
    """Extract the JSON object inside ```json ...``` from a string."""
    if not isinstance(text, str):
        return None
    m = FENCE_RE.search(text)
    if not m:
        return None
    blob = m.group(1).strip()
    try:
        return json.loads(blob)
    except json.JSONDecodeError:
        return None

def extract_reason(text: str) -> str | None:
    """
    Extract the 'reason' field from the JSON embedded in the 'response' string.
    Returns None if not found / not parseable.
    """
    obj = extract_json_from_response(text)
    if isinstance(obj, dict) and "reason" in obj and isinstance(obj["reason"], str):
        return obj["reason"].strip()
    return None

# Example filename:
# GPT-score_Claude_answers_category_1.2_HIV_EQ.json
# GPT-score_Llama-1B_answers_category_1.4_HIV_EQ.json
NAME_RE = re.compile(r"answers_category_(\d+)\.(\d+)_", re.IGNORECASE)

def parse_category_iteration_from_filename(path: str) -> tuple[int, int] | None:
    """Return (category_id, iteration_number) from filename, or None."""
    base = os.path.basename(path)
    m = NAME_RE.search(base)
    if not m:
        return None
    return int(m.group(1)), int(m.group(2))

def norm(s: str) -> str:
    return re.sub(r"[^a-z0-9]+", "", s.lower())

# ----------------------------
# 3) Build a mapping: folder_name -> df.model value
#    (Auto-guess; you can override with a manual dict if needed)
# ----------------------------
models_root = "."  # your folders are in the current directory per your listing
df_models = sorted(df["model"].dropna().unique().tolist())
df_models_norm = {m: norm(m) for m in df_models}

# List folders in root that might contain GPT-score_*.json
candidate_dirs = [
    d for d in os.listdir(models_root)
    if os.path.isdir(os.path.join(models_root, d))
]

folder_to_model = {}
unmatched_folders = []

for folder in candidate_dirs:
    # Only consider folders that actually have GPT-score jsons somewhere inside
    folder_path = os.path.join(models_root, folder)
    has_json = bool(glob.glob(os.path.join(folder_path, "**", "GPT-score_*.json"), recursive=True))
    if not has_json:
        continue

    f_norm = norm(folder)

    # best match: folder name contained in model name (or vice versa)
    best = None
    for m in df_models:
        mn = df_models_norm[m]
        if f_norm and (f_norm in mn or mn in f_norm):
            best = m
            break

    if best is None:
        unmatched_folders.append(folder)
    else:
        folder_to_model[folder] = best

# OPTIONAL: If any folders didn't match automatically, set them manually here.
# Example:
folder_to_model.update({
     "Claude": "Claude 3.5 Sonnet",
     "Gemini_2.5Pro": "Gemini 2.5 Pro",
     "Llama_cluster": "Llama 3.3-70B-Instruct",
     "Llama-1B": "Llama 3.2-1B-Instruct",
     "Llama-8B": "Llama 3.1-8B-Instruct",
     "Gemma-3-27B": "Gemma 3 27B",
     "MedGemma-3-27B": "MedGemma 27B",
     "Med42": "Med42-70B",
     "Meditron_cluster": "Meditron 3-70B",
     "NVLM": "NVLM-70B",
     "MedGemma-3-27B-RAG": "MedGemma 27B RAG",
     "Llama-8B-RAG": "Llama 3.1-8B-Instruct RAG",
 })


print("Auto folder_to_model mapping:")
for k, v in folder_to_model.items():
    print(f"  {k} -> {v}")

if unmatched_folders:
    print("\nUnmatched folders (set these in folder_to_model.update(...) above if they contain scoring jsons):")
    for f in unmatched_folders:
        print(" ", f)

# ----------------------------
# 4) Read all GPT-score jsons and build lookup:
#    (model, category_id, iteration_number, question_index) -> reason
# ----------------------------
reason_lookup = {}

for folder, model_name in folder_to_model.items():
    folder_path = os.path.join(models_root, folder)
    json_paths = glob.glob(os.path.join(folder_path, "**", "GPT-score_*.json"), recursive=True)

    for jp in json_paths:
        ci = parse_category_iteration_from_filename(jp)
        if ci is None:
            continue
        category_id, iteration_number = ci

        try:
            with open(jp, "r", encoding="utf-8") as f:
                data = json.load(f)
        except Exception:
            continue

        # Your files are like: [ {"response": "...json..."}, {"response": "..."} , ...]
        if not isinstance(data, list):
            continue

        for i, item in enumerate(data, start=1):
            # i is the question_index within that category/iteration file
            if not isinstance(item, dict):
                continue
            reason = extract_reason(item.get("response"))
            if reason:
                reason_lookup[(model_name, category_id, iteration_number, i)] = reason

print(f"\nBuilt reason_lookup entries: {len(reason_lookup):,}")

# ----------------------------
# 4b) Read REPHRASED GOLD scoring jsons
# ----------------------------
import glob
import json

gold_jsons = glob.glob(
    os.path.join(GOLD_BASE_DIR, "GPT-score_rephrased_true_answers_answers_category_*.json")
)

print(f"Found {len(gold_jsons)} rephrased gold json files")

for jp in gold_jsons:
    ci = parse_category_iteration_from_filename(jp)
    if ci is None:
        continue

    category_id, iteration_number = ci

    try:
        with open(jp, "r", encoding="utf-8") as f:
            data = json.load(f)
    except Exception:
        continue

    if not isinstance(data, list):
        continue

    for i, item in enumerate(data, start=1):
        if not isinstance(item, dict):
            continue

        reason = extract_reason(item.get("response"))
        if reason:
            reason_lookup[
                (GOLD_MODEL_NAME, category_id, iteration_number, i)
            ] = reason

# ----------------------------
# 5) Fill df['reasoning'] by matching each row
# ----------------------------
keys = list(
    zip(df["model"], df["category_id"], df["iteration_number"], df["question_index"])
)
df["reasoning"] = [reason_lookup.get(k, pd.NA) for k in keys]

# ----------------------------
# 6) Save output
# ----------------------------
out_path = "all_questions_answers_scores_plus_reason.csv"
df.to_csv(out_path, index=False)
print(f"\nSaved: {out_path}")

# OPTIONAL sanity checks:
print("\nReasoning filled count:", df["reasoning"].notna().sum(), "/", len(df))
print("Examples (non-null):")
print(df.loc[df["reasoning"].notna(), ["model","category_id","iteration_number","question_index","reasoning"]].head(3))
````

````python
import os
import re
import json
import glob
import pandas as pd

# ----------------------------
# 1) Load CSV
# ----------------------------
df = pd.read_csv("all_questions_answers_scores.csv")

# Insert "reasoning" right after MedGPT5
insert_at = df.columns.get_loc("MedGPT5") + 1
if "reasoning" not in df.columns:
    df.insert(insert_at, "reasoning", pd.NA)

# ----------------------------
# 2) Paths / constants
# ----------------------------
models_root = "/cluster/home/gcardenal/HIV/medical_llm_evaluation/deploy_medical_llm_evaluation/model_answers/results-GPT-score"  # your notebook working dir: model_answers/results-GPT-score/

GOLD_MODEL_NAME = "Rephrased Gold Answers"
GOLD_BASE_DIR = (
    "/cluster/home/gcardenal/HIV/medical_llm_evaluation/"
    "deploy_medical_llm_evaluation/gold_standard_rephrasing/"
    "results-GPT-score-rephrased"
)

# ----------------------------
# 3) Helpers
# ----------------------------
FENCE_RE = re.compile(r"```(?:json)?\s*(\{.*?\})\s*```", re.DOTALL | re.IGNORECASE)
NAME_RE = re.compile(r"answers_category_(\d+)\.(\d+)_", re.IGNORECASE)

def extract_json_from_response(text: str):
    if not isinstance(text, str):
        return None
    m = FENCE_RE.search(text)
    if not m:
        return None
    try:
        return json.loads(m.group(1).strip())
    except json.JSONDecodeError:
        return None

def extract_reason(text: str):
    obj = extract_json_from_response(text)
    if isinstance(obj, dict) and isinstance(obj.get("reason"), str):
        return obj["reason"].strip()
    return None

def parse_category_iteration_from_filename(path: str):
    base = os.path.basename(path)
    m = NAME_RE.search(base)
    if not m:
        return None
    return int(m.group(1)), int(m.group(2))

def norm(s: str) -> str:
    return re.sub(r"[^a-z0-9]+", "", s.lower())

# ----------------------------
# 4) Map folders -> df.model names
# ----------------------------
folder_to_model = {
    "Claude": "Claude 3.5 Sonnet",
    "Gemini_2.5Pro": "Gemini 2.5 Pro",
    "Llama_cluster": "Llama 3.3-70B-Instruct",
    "Llama-1B": "Llama 3.2-1B-Instruct",
    "Llama-8B": "Llama 3.1-8B-Instruct",
    "Gemma-3-27B": "Gemma 3 27B",
    "MedGemma-3-27B": "MedGemma 27B",
    "Med42": "Med42-70B",
    "Meditron_cluster": "Meditron 3-70B",
    "NVLM": "NVLM-70B",
    "MedGemma-3-27B-RAG": "MedGemma 27B RAG",
    "Llama-8B-RAG": "Llama 3.1-8B-Instruct RAG",
}

# ----------------------------
# 5) Build mapping:
#    (model, category_id, iteration_number) -> list_of_reasons_in_order
#    + diagnostics: (model, cat, iter, json_path) -> list of entry indices where reason is None
# ----------------------------
reasons_by_group = {}
none_reason_entries = {}  # key: (model, category, iter, path) -> [entry_idx, ...]

def add_file_reasons(model_name: str, json_path: str):
    ci = parse_category_iteration_from_filename(json_path)
    if ci is None:
        return
    category_id, iteration_number = ci

    try:
        with open(json_path, "r", encoding="utf-8") as f:
            data = json.load(f)
    except Exception:
        return

    if not isinstance(data, list):
        return

    reasons = []
    none_idxs = []

    for idx, item in enumerate(data):
        if not isinstance(item, dict):
            reasons.append(None)
            none_idxs.append(idx)
            continue

        r = extract_reason(item.get("response"))
        reasons.append(r)  # keep position even if None (preserves alignment)

        if r is None:
            none_idxs.append(idx)

    reasons_by_group[(model_name, category_id, iteration_number)] = reasons

    if none_idxs:
        none_reason_entries[(model_name, category_id, iteration_number, json_path)] = none_idxs

# --- regular model folders ---
for folder, model_name in folder_to_model.items():
    folder_path = os.path.join(models_root, folder)
    json_paths = glob.glob(os.path.join(folder_path, "**", "GPT-score_*.json"), recursive=True)
    for jp in json_paths:
        add_file_reasons(model_name, jp)

# --- gold rephrased folder ---
gold_paths = glob.glob(
    os.path.join(GOLD_BASE_DIR, "GPT-score_rephrased_true_answers_answers_category_*.json")
)
for jp in gold_paths:
    add_file_reasons(GOLD_MODEL_NAME, jp)

print(f"Loaded reason lists for groups: {len(reasons_by_group):,}")

# ----------------------------
# 6) Fill df['reasoning'] by group order (sorted by question_index)
#    + track which df rows were expected to have reasoning (group exists)
# ----------------------------
df["reasoning"] = pd.NA  # reset in case rerun

expected_reasoning_rows = set()
filled = 0

for (model_name, category_id, iteration_number), idxs in df.groupby(
    ["model", "category_id", "iteration_number"]
).groups.items():

    reasons = reasons_by_group.get((model_name, int(category_id), int(iteration_number)))
    if not reasons:
        continue

    # these rows correspond to a group that exists in JSON => "expected"
    expected_reasoning_rows.update(idxs)

    # assign by CSV question order
    group_index_list = list(idxs)
    group_sorted = sorted(group_index_list, key=lambda i: df.at[i, "question_index"])

    n = min(len(group_sorted), len(reasons))
    for k in range(n):
        r = reasons[k]
        if r:  # only fill if we successfully extracted a reason
            df.at[group_sorted[k], "reasoning"] = r
            filled += 1

print("Reasoning filled cells:", filled)

# ----------------------------
# 6b) Diagnostics
# ----------------------------

# A) Rows that belong to a group with JSONs but still have null reasoning
null_expected_mask = df.index.isin(expected_reasoning_rows) & df["reasoning"].isna()
null_expected_df = df.loc[
    null_expected_mask,
    ["model", "category_id", "iteration_number", "question_index"]
].sort_values(["model", "category_id", "iteration_number", "question_index"])

print("\nNull reasoning rows (group matched but no reason extracted):")
print(f"{len(null_expected_df)} rows")
print(null_expected_df.head(50))

# B) Rows with NO matching JSON group at all
all_group_keys = set(reasons_by_group.keys())
df_group_keys = set(
    (row.model, int(row.category_id), int(row.iteration_number))
    for row in df.itertuples()
)
unmatched_groups = df_group_keys - all_group_keys

unmatched_mask = df.apply(
    lambda r: (r.model, int(r.category_id), int(r.iteration_number)) in unmatched_groups,
    axis=1,
)
unmatched_df = df.loc[
    unmatched_mask,
    ["model", "category_id", "iteration_number", "question_index"]
].sort_values(["model", "category_id", "iteration_number", "question_index"])

print("\nRows with NO matching JSON group at all:")
print(f"{len(unmatched_df)} rows")
print(unmatched_df.head(50))

# C) Which JSON entries had None reasons (per file)
print("\nJSON entries with None reason (by file):")
if not none_reason_entries:
    print("None")
else:
    # show a compact summary
    total_none = sum(len(v) for v in none_reason_entries.values())
    print(f"Files with >=1 None reason: {len(none_reason_entries)}")
    print(f"Total None entries across those files: {total_none}")

    # print first N files (adjust as needed)
    N = 20
    shown = 0
    for (model_name, cat, it, path), idx_list in sorted(
        none_reason_entries.items(),
        key=lambda kv: (kv[0][0], kv[0][1], kv[0][2], kv[0][3]),
    ):
        print(f"- {model_name} | cat={cat} iter={it} | none_count={len(idx_list)} | file={path}")
        print(f"  entry_indices (0-based): {idx_list[:50]}" + (" ..." if len(idx_list) > 50 else ""))
        shown += 1
        if shown >= N:
            remaining = len(none_reason_entries) - N
            if remaining > 0:
                print(f"... ({remaining} more files not shown)")
            break

# ----------------------------
# 7) Save output
# ----------------------------
out_path = "all_questions_answers_scores_plus_reason.csv"
df.to_csv(out_path, index=False)
print("\nSaved:", out_path)

# quick sanity:
print("Non-null reasoning:", df["reasoning"].notna().sum(), "/", len(df))
print(
    df.loc[
        df["reasoning"].notna(),
        ["model", "category_id", "iteration_number", "question_index", "reasoning"],
    ].head(3)
)
````

````python
def toks(s):
    return set(re.findall(r"[a-zA-Z0-9]+", str(s).lower()))

def overlap(a, b):
    A, B = toks(a), toks(b)
    return 0 if not A or not B else len(A & B) / len(A | B)

for (m, c, it), idxs in df.groupby(["model","category_id","iteration_number"]).groups.items():
    idxs = sorted(list(idxs), key=lambda i: df.at[i, "question_index"])
    if len(idxs) < 10:
        continue

    # pick the column that has the model answer you want to compare against
    ans_col = "model_answer"  # change to your real column name

    scores0, scoresP1, scoresM1 = [], [], []
    for j, i in enumerate(idxs):
        r = df.at[i, "reasoning"]
        if pd.isna(r):
            continue

        a0 = df.at[i, ans_col]
        s0 = overlap(r, a0)
        scores0.append(s0)

        if j + 1 < len(idxs):
            scoresP1.append(overlap(r, df.at[idxs[j+1], ans_col]))
        if j - 1 >= 0:
            scoresM1.append(overlap(r, df.at[idxs[j-1], ans_col]))

    if scores0:
        m0 = sum(scores0)/len(scores0)
        mp = sum(scoresP1)/len(scoresP1) if scoresP1 else 0
        mm = sum(scoresM1)/len(scoresM1) if scoresM1 else 0

        if mp > m0 + 0.02 or mm > m0 + 0.02:
            print("POSSIBLE SHIFT:", (m,c,it), "m0=", round(m0,3), "m+1=", round(mp,3), "m-1=", round(mm,3))
````

````python
import pandas as pd

df_new = pd.read_csv("all_questions_answers_scores_plus_reason.csv")
df_old = pd.read_csv("./all_questions_answers_scores_plus_reason_old.csv")

KEY_COLS = ["model", "category_id", "iteration_number", "question_index"]

def normalize(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df["model"] = df["model"].astype(str).str.strip()
    for c in ["category_id", "iteration_number", "question_index"]:
        df[c] = pd.to_numeric(df[c], errors="coerce")
    return df

df_new_n = normalize(df_new)
df_old_n = normalize(df_old)

# ---- FIX: old question_index is 1-based, new is 0-based ----
df_old_n["question_index"] = df_old_n["question_index"] - 1

# (Optional) If you want to be strict, drop any rows that became invalid
df_old_n = df_old_n.dropna(subset=KEY_COLS)

# Filter NEW by keys present in OLD
old_keys = df_old_n[KEY_COLS].drop_duplicates()

df_filtered = df_new_n.merge(
    old_keys,
    on=KEY_COLS,
    how="inner",
)

out_path = "all_questions_answers_scores_plus_reason_filtered.csv"
df_filtered.to_csv(out_path, index=False)

print("Old rows:", len(df_old), " | unique keys:", len(old_keys))
print("New rows:", len(df_new))
print("Rows kept:", len(df_filtered), "/", len(df_new))
print("Saved:", out_path)

# Extra sanity: check if any OLD keys still missing in NEW after fix
missing = old_keys.merge(
    df_new_n[KEY_COLS].drop_duplicates(),
    on=KEY_COLS,
    how="left",
    indicator=True,
)
missing = missing[missing["_merge"] == "left_only"].drop(columns=["_merge"])
print("Unique OLD keys still missing in NEW after index shift:", len(missing))
if len(missing) > 0:
    print(missing.head(20))
````

````python
import pandas as pd
import re
import unicodedata

df_reason = pd.read_csv("all_questions_answers_scores_plus_reason.csv")
df_questions = pd.read_csv("./questions.csv")

assert "question" in df_reason.columns, "df_reason missing column: question"
assert "question" in df_questions.columns, "df_questions missing column: question"

def norm_str(s: str) -> str:
    if not isinstance(s, str):
        return ""
    # normalize unicode (quotes, accents, etc.)
    s = unicodedata.normalize("NFKC", s)

    s = s.lower().strip()
    s = re.sub(r"\s+", " ", s)  # collapse whitespace

    # remove common leading labels like "q12:", "question 12:", "12)"
    s = re.sub(r"^(question|q)\s*\d+\s*[:\)\.\-]\s*", "", s)
    s = re.sub(r"^\d+\s*[:\)\.\-]\s*", "", s)

    # remove punctuation-ish (keep alnum and spaces)
    s = re.sub(r"[^a-z0-9\s]+", "", s)
    s = re.sub(r"\s+", " ", s).strip()
    return s

df_reason["_q_norm"] = df_reason["question"].apply(norm_str)
df_questions["_q_norm"] = df_questions["question"].apply(norm_str)

valid = set(df_questions["_q_norm"].dropna())
print("questions.csv rows:", len(df_questions))
print("questions.csv unique normalized:", len(valid))

# overlap diagnostics
reason_unique = set(df_reason["_q_norm"].dropna())
print("df_reason rows:", len(df_reason))
print("df_reason unique normalized:", len(reason_unique))
print("normalized intersection:", len(valid & reason_unique))

# filter (keep ONLY those whose normalized question exists in questions.csv)
df_filtered = df_reason[df_reason["_q_norm"].isin(valid)].copy()
df_filtered.drop(columns=["_q_norm"], inplace=True)

out_path = "all_questions_answers_scores_plus_reason_filtered.csv"
df_filtered.to_csv(out_path, index=False)

print("Filtered rows:", len(df_filtered))
print("Saved:", out_path)
````
