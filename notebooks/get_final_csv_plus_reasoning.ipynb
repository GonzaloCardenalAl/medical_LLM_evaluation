````python
import os
import re
import json
import glob
import pandas as pd

# ----------------------------
# 1) Load CSV
# ----------------------------
df = pd.read_csv("all_questions_answers_scores.csv")

# Insert "reasoning" right after MedGPT5
insert_at = df.columns.get_loc("MedGPT5") + 1
if "reasoning" not in df.columns:
    df.insert(insert_at, "reasoning", pd.NA)

# ----------------------------
# 2) Paths / constants
# ----------------------------
models_root = "/cluster/home/gcardenal/HIV/medical_llm_evaluation/deploy_medical_llm_evaluation/model_answers/results-GPT-score"  # your notebook working dir: model_answers/results-GPT-score/

GOLD_MODEL_NAME = "Rephrased Gold Answers"
GOLD_BASE_DIR = (
    "/cluster/home/gcardenal/HIV/medical_llm_evaluation/"
    "deploy_medical_llm_evaluation/gold_standard_rephrasing/"
    "results-GPT-score-rephrased"
)

# ----------------------------
# 3) Helpers
# ----------------------------
FENCE_RE = re.compile(r"```(?:json)?\s*(\{.*?\})\s*```", re.DOTALL | re.IGNORECASE)
NAME_RE = re.compile(r"answers_category_(\d+)\.(\d+)_", re.IGNORECASE)

def extract_json_from_response(text: str):
    if not isinstance(text, str):
        return None
    m = FENCE_RE.search(text)
    if not m:
        return None
    try:
        return json.loads(m.group(1).strip())
    except json.JSONDecodeError:
        return None

def extract_reason(text: str):
    obj = extract_json_from_response(text)
    if isinstance(obj, dict) and isinstance(obj.get("reason"), str):
        return obj["reason"].strip()
    return None

def parse_category_iteration_from_filename(path: str):
    base = os.path.basename(path)
    m = NAME_RE.search(base)
    if not m:
        return None
    return int(m.group(1)), int(m.group(2))

def norm(s: str) -> str:
    return re.sub(r"[^a-z0-9]+", "", s.lower())

# ----------------------------
# 4) Map folders -> df.model names
# ----------------------------
folder_to_model = {
    "Claude": "Claude 3.5 Sonnet",
    "Gemini_2.5Pro": "Gemini 2.5 Pro",
    "Llama_cluster": "Llama 3.3-70B-Instruct",
    "Llama-1B": "Llama 3.2-1B-Instruct",
    "Llama-8B": "Llama 3.1-8B-Instruct",
    "Gemma-3-27B": "Gemma 3 27B",
    "MedGemma-3-27B": "MedGemma 27B",
    "Med42": "Med42-70B",
    "Meditron_cluster": "Meditron 3-70B",
    "NVLM": "NVLM-70B",
    "MedGemma-3-27B-RAG": "MedGemma 27B RAG",
    "Llama-8B-RAG": "Llama 3.1-8B-Instruct RAG",
}

# ----------------------------
# 5) Build mapping:
#    (model, category_id, iteration_number) -> list_of_reasons_in_order
#    + diagnostics: (model, cat, iter, json_path) -> list of entry indices where reason is None
# ----------------------------
reasons_by_group = {}
none_reason_entries = {}  # key: (model, category, iter, path) -> [entry_idx, ...]

def add_file_reasons(model_name: str, json_path: str):
    ci = parse_category_iteration_from_filename(json_path)
    if ci is None:
        return
    category_id, iteration_number = ci

    try:
        with open(json_path, "r", encoding="utf-8") as f:
            data = json.load(f)
    except Exception:
        return

    if not isinstance(data, list):
        return

    reasons = []
    none_idxs = []

    for idx, item in enumerate(data):
        if not isinstance(item, dict):
            reasons.append(None)
            none_idxs.append(idx)
            continue

        r = extract_reason(item.get("response"))
        reasons.append(r)  # keep position even if None (preserves alignment)

        if r is None:
            none_idxs.append(idx)

    reasons_by_group[(model_name, category_id, iteration_number)] = reasons

    if none_idxs:
        none_reason_entries[(model_name, category_id, iteration_number, json_path)] = none_idxs

# --- regular model folders ---
for folder, model_name in folder_to_model.items():
    folder_path = os.path.join(models_root, folder)
    json_paths = glob.glob(os.path.join(folder_path, "**", "GPT-score_*.json"), recursive=True)
    for jp in json_paths:
        add_file_reasons(model_name, jp)

# --- gold rephrased folder ---
gold_paths = glob.glob(
    os.path.join(GOLD_BASE_DIR, "GPT-score_rephrased_true_answers_answers_category_*.json")
)
for jp in gold_paths:
    add_file_reasons(GOLD_MODEL_NAME, jp)

print(f"Loaded reason lists for groups: {len(reasons_by_group):,}")

# ----------------------------
# 6) Fill df['reasoning'] by group order (sorted by question_index)
#    + track which df rows were expected to have reasoning (group exists)
# ----------------------------
df["reasoning"] = pd.NA  # reset in case rerun

expected_reasoning_rows = set()
filled = 0

for (model_name, category_id, iteration_number), idxs in df.groupby(
    ["model", "category_id", "iteration_number"]
).groups.items():

    reasons = reasons_by_group.get((model_name, int(category_id), int(iteration_number)))
    if not reasons:
        continue

    # these rows correspond to a group that exists in JSON => "expected"
    expected_reasoning_rows.update(idxs)

    # assign by CSV question order
    group_index_list = list(idxs)
    group_sorted = sorted(group_index_list, key=lambda i: df.at[i, "question_index"])

    n = min(len(group_sorted), len(reasons))
    for k in range(n):
        r = reasons[k]
        if r:  # only fill if we successfully extracted a reason
            df.at[group_sorted[k], "reasoning"] = r
            filled += 1

print("Reasoning filled cells:", filled)

# ----------------------------
# 6b) Diagnostics
# ----------------------------

# A) Rows that belong to a group with JSONs but still have null reasoning
null_expected_mask = df.index.isin(expected_reasoning_rows) & df["reasoning"].isna()
null_expected_df = df.loc[
    null_expected_mask,
    ["model", "category_id", "iteration_number", "question_index"]
].sort_values(["model", "category_id", "iteration_number", "question_index"])

print("\nNull reasoning rows (group matched but no reason extracted):")
print(f"{len(null_expected_df)} rows")
print(null_expected_df.head(50))

# B) Rows with NO matching JSON group at all
all_group_keys = set(reasons_by_group.keys())
df_group_keys = set(
    (row.model, int(row.category_id), int(row.iteration_number))
    for row in df.itertuples()
)
unmatched_groups = df_group_keys - all_group_keys

unmatched_mask = df.apply(
    lambda r: (r.model, int(r.category_id), int(r.iteration_number)) in unmatched_groups,
    axis=1,
)
unmatched_df = df.loc[
    unmatched_mask,
    ["model", "category_id", "iteration_number", "question_index"]
].sort_values(["model", "category_id", "iteration_number", "question_index"])

print("\nRows with NO matching JSON group at all:")
print(f"{len(unmatched_df)} rows")
print(unmatched_df.head(50))

# C) Which JSON entries had None reasons (per file)
print("\nJSON entries with None reason (by file):")
if not none_reason_entries:
    print("None")
else:
    # show a compact summary
    total_none = sum(len(v) for v in none_reason_entries.values())
    print(f"Files with >=1 None reason: {len(none_reason_entries)}")
    print(f"Total None entries across those files: {total_none}")

    # print first N files (adjust as needed)
    N = 20
    shown = 0
    for (model_name, cat, it, path), idx_list in sorted(
        none_reason_entries.items(),
        key=lambda kv: (kv[0][0], kv[0][1], kv[0][2], kv[0][3]),
    ):
        print(f"- {model_name} | cat={cat} iter={it} | none_count={len(idx_list)} | file={path}")
        print(f"  entry_indices (0-based): {idx_list[:50]}" + (" ..." if len(idx_list) > 50 else ""))
        shown += 1
        if shown >= N:
            remaining = len(none_reason_entries) - N
            if remaining > 0:
                print(f"... ({remaining} more files not shown)")
            break

# ----------------------------
# 7) Save output
# ----------------------------
out_path = "all_questions_answers_scores_plus_reason.csv"
df.to_csv(out_path, index=False)
print("\nSaved:", out_path)

# quick sanity:
print("Non-null reasoning:", df["reasoning"].notna().sum(), "/", len(df))
print(
    df.loc[
        df["reasoning"].notna(),
        ["model", "category_id", "iteration_number", "question_index", "reasoning"],
    ].head(3)
)
````
